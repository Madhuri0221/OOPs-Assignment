```python
# 10) What are class and static methods in Python?

"""
The class and static methods are the inbuit methods used in python
A class method takes cls as an input argument instead of taking self as the argument
The cls in the return statement automatically invokes __init__and the object is created
so we need not to manually create an object
Also the cls in the input argument is like self
this cls lets us directly connect with class
we can declare class method foe external functions also

lets see some exmples

"""
class Student:
  def __init__(self,name,age):
    self.name = name
    self.age = age
  @classmethod
  def details(cls,name,age):
    return cls(name,age)

s=Student.details("Madrid",22)
print(s.name,s.age)

"""
Making external function as class method

Student.external_method=classmethod(external_method)

"""

def ext_method(cls,degree):
  return cls(degree)

class A:
  def __init__(self,degree):
    self.degree = degree

A.ext_method = classmethod(ext_method)
a=A.ext_method("Data Analytics")
print(a.degree)

"""
Static method: The static method is same as function
No need to create an object
doesnot take self or cls as input argument
We cant modify method
It doesnot access the attributes or methods of a class

"""
class Calculator:

	@staticmethod
	def add(x,y):
		return x+y

	@staticmethod
	def subtract(x,y):
		return x-y

Calculator.add(2,3)
Calculator.add(10,8)
```

```python
#  13) What is a property decorator in Python?
"""
A property decorator is the built-in decorator in python
Using this or applying this to a method lets us access the method through object
as an attribute
Which shows clean code
Lets say we have applied @property to a method
now, we can use that method as an attribute and set,get,del the values/methods
ie., we can use setter,getter,deleter decorators
While coding we no need to explicitly use getter method because @property itself
acts as getter
We use deleter when we want to delete an object attribute

"""

class A:

  def __init__(self,r):
    self.r = r

  @property
  def radius(self):
    print("Radius:",self.r)

  @radius.setter
  def radius(self,r_new):
    self.r = r_new

  @radius.deleter
  def radius(self):
    del self.r

a=A(2)
print("Old radius")
a.radius
print("")
a.radius = 5
print("Updated radius")
a.radius
print("")
print("Deleting the radius attribute")
del a.radius
print(a.radius)

"""
So from the above example we have used all the decorators that can be used
when we apply @property decorator to the method

"""
```

```python
# 15) What is an abstract class in Python?

"""
An abstract class is the blueprint for other classes
from abc we can import ABC,abstractmethod
ABC is used as an input parameter as the base class for a class
ABC - Abstract Base Class
abstarctmethod is applied to the method of the abstarct class

"""

from abc import ABC,abstractmethod

class Animal(ABC):
  @abstractmethod
  def sound(self):
    pass

class Dog(Animal):
  def sound(self):
    print("Bark!")

class Cat(Animal):
  def sound(self):
    print("Meow!")

def making_sounds(obj):
  obj.sound()

c=Cat()
d=Dog()

making_sounds(d)
```

```python
# 24) What is method chaining in Python OOP?

"""
Method chaining means multiple methods are called on the same object
in a single line

Each method returns the object

Advantages: Increases readability, efficient execution

"""

class Calculator:
    def __init__(self, value=0):
        self.value = value

    def add(self, num):
        self.value += num
        return self  # Returning the object itself

    def multiply(self, num):
        self.value *= num
        return self  # Returning the object itself

    def result(self):
        return self.value

# Using method chaining
calc = Calculator()
final_value = calc.add(5).multiply(3).result()

print(final_value)
```

```python
# 25) What is the purpose of the __call__ method in Python?

"""
__call__is the special method used to call the object as how we call a method

"""

class Multiplier:
    def __init__(self, factor):
        self.factor = factor

    def __call__(self, number):
        return number * self.factor

# Creating an object
double = Multiplier(2)

# Calling the object like a function
result = double(5)  # Equivalent to double.__call__(5)

print(result)
```

```python
# 2) Write a program to create an abstract class Shape with a method area(). Derive classes Circle and Rectangle
# from it and implement the area() method in both

from abc import abstractmethod
class Shape:
  @abstractmethod
  def area():
    pass

class Circle(Shape):

  def __init__(self,r):
    self.r = r

  def area(self):
    print(f"Area of circle is: {3.14*self.r**2}")

class Rectangle(Shape):

  def __init__(self,l,b):
    self.l = l
    self.b = b

  def area(self):
    print(f"Area of rectangle is: {self.l*self.b}")


c=Circle(5)
c.area()

r=Rectangle(2,3)
r.area()
```

```python
#  5) Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes
# balance and methods to deposit, withdraw, and check balance.

class BankAccount:

  def __init__(self,balance):
    self.__balance = balance

  def deposit(self,amount):
    print(f"Opening balance: {self.__balance}")
    print("")
    print(f"Before depositing: {self.__balance}")
    if amount > 0:
      self.__balance += amount
      print(f"After depositing: {self.__balance}")
    else:
      print("Deposit positive amount")

  def withdraw(self,amount):
    print(f"Before withdrawing: {self.__balance}")
    if amount > 0 and amount <= self.__balance:
      self.__balance -= amount
      print(f"After withdrawing: {self.__balance}")
    else:
      print("Provide positive number/Insufficient funds")


b=BankAccount(5000)
b.deposit(1000)
b.withdraw(2000)
print("")
print("Accessing private attribute outside the class and changing its value")
print("")
b._BankAccount__balance = 10000
b.deposit(500)


```

```python
# 6) Demonstrate runtime polymorphism using a method play() in a base class Instrument. Derive classes Guitar
# and Piano that implement their own version of play().

class Instrument:
  def play(self):
    print("Base")

class Guitar(Instrument):
  def play(self):
    print("Playing Guitar")

class Piano(Instrument):
  def play(self):
    print("Playing Piano")

def perform(instrument):
  instrument.play()

g=Guitar()
p=Piano()

perform(g)
perform(p)
```

```python
# 7)  Create a class MathOperations with a class method add_numbers() to add two numbers and a static
# method subtract_numbers() to subtract two numbers.

class MathOperations:

  def __init__(self,result):
    self.result = result

  @classmethod
  def add_numbers(cls,a,b):
    return cls(a+b)

  @staticmethod
  def subtract_numbers(n1,n2):
    if n1>n2:
      print(n1-n2)
    else:
      print("n1 should be greater than n2 to get a positive output")

x = MathOperations.add_numbers(2,3)
print("Invoking classmethod")
print("Sum of numbers:", x.result)
print("")
print("Invoking static method")
MathOperations.subtract_numbers(10,5)

```

```python
# 9)  Write a class Fraction with attributes numerator and denominator. Override the str method to display the
# fraction as "numerator/denominator"

class Fraction:

  def __init__(self,numerator,denominator):
    self.numerator = numerator
    self.denominator = denominator

  def __str__(self):
    return f"{self.numerator}/{self.denominator}"

f=Fraction(3,4)
print("Applying print to the object:",f)
print("Applying str to the object, to get the string format")
str(f)

```

```python
# 10) Demonstrate operator overloading by creating a class Vector and overriding the add method to add two
# vectors

class Vector:

  def __init__(self,n1,n2):
    self.n1 = n1
    self.n2 = n2

  def __add__(self,other):
    return self.n1+other.n1,self.n2+other.n2

v1=Vector(1,2)
v2=Vector(2,3)
result = v1+v2
print(result)
```

```python
#  12)  Implement a class Student with attributes name and grades. Create a method average_grade() to compute
# the average of the grades.

class Student:

  def __init__(self,name,grades):
    self.name = name
    self.grades = grades

  def average_grade(self):
    if self.grades:
      average = sum(self.grades)/len(self.grades)
      return average
    else:
      print("Grades are not given, please provide grades")

s1=Student("Alice",[50,76,35])
s1.average_grade()
```

```python
# 14) Create a class Employee with a method calculate_salary() that computes the
# salary based on hours worked and hourly rate. Create a derived class Manager that adds a bonus to the salary

class Employee:

  def __init__(self,salary,hours_worked,hourly_rate):
    self.salary = salary
    self.hours_worked = hours_worked
    self.hourly_rate = hourly_rate
    self.total_salary = 0

  def calculate_salary(self):
    self.total_salary = self.salary * self.hours_worked * self.hourly_rate
    return self.total_salary

class Manager(Employee):

  def __init__(self,salary,hours_worked,hourly_rate,bonus):
    super().__init__(salary,hours_worked,hourly_rate)
    self.bonus = bonus

  def bonus_received(self):
      if self.total_salary > 200:
        print(f"Total + Bonus: {self.total_salary + self.bonus}")
      else:
        print(f"Bonus not received, only salary is credited: {self.total_salary}")

m1=Manager(100,4,2,100)
m1.calculate_salary()
m1.bonus_received()

m2=Manager(10,4,2,100)
m2.calculate_salary()
m2.bonus_received()
```

```python
#  15) Create a class Product with attributes name, price, and quantity. Implement a method total_price() that
# calculates the total price of the product.

class Product:

  total_cost = 0

  def __init__(self,name,price,quantity):
    self.name = name
    self.price = price
    self.quantity = quantity
    # self.total_cost = 0

  def total_price(self):
    total_cost = self.price*self.quantity
    if total_cost > 0:
      print(f"Total price of the {self.name} of quantity {self.quantity} is: {total_cost}")
    else:
      print(f"Cost / Quantity of {self.name} will not be negative / Zero, please check")

p=Product("Ghee",100,2)
p.total_price()

o=Product("Rice",-1500,3)
o.total_price()

q=Product("Milk",50,-5)
q.total_price()

q1=Product("Curd",50,0)
q1.total_price()
```

```python
#  16) Create a class Animal with an abstract method sound(). Create two derived classes Cow and Sheep that
# implement the sound() method.

from abc import ABC,abstractmethod
class Animal(ABC):
  @abstractmethod
  def sound(self):
    pass

class Cow(Animal):
  def sound(self):
    print("Cow sounds: Mooo")

class Sheep(Animal):
  def sound(self):
    print("Sheep sounds: Baaa")

def making_sound(obj):
  obj.sound()

c=Cow()
s=Sheep()

making_sound(c)

"""
I have implemented runtime polymorphism
We can give object 's' to check whats inside the method

"""
```

```python
#  17) Create a class Book with attributes title, author, and year_published. Add a method get_book_info() that
# returns a formatted string with the book's details.

class Book:

  def __init__(self,title,author,year_published):
    self.title = title
    self.author = author
    self.year_published = year_published

  def get_book_info(self):
    print(f"Book details:\n Title:{self.title}\n Author:{self.author}\n Year_published:{self.year_published}")

b=Book("The Hidden Life of Trees","Peter Wohlleben",2015)
b.get_book_info()
```

```python
# 18) Create a class House with attributes address and price. Create a derived class Mansion that adds an
# attribute number_of_rooms.

class House:

  def __init__(self,address,price):
    self.address = address
    self.price = price

  def details(self):
    print(f"Details of house:\n Address:{self.address}\n Price:{self.price}")

class Mansion(House):

  def __init__(self,address,price,number_of_rooms):
    super().__init__(address,price)
    self.number_of_rooms = number_of_rooms

  def details_including_rooms(self):
    print(f"Details of house including rooms:\n Address: {self.address}\n Price: {self.price}\n Number of rooms: {self.number_of_rooms}")

m=Mansion("Plot no.32A,Jubliee Hills",10000000,4)
m.details_including_rooms()
```
